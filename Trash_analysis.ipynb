```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
def read_csv(path):
  df = pd.read_csv(path)
  file_name =os.path.basename(path)
  #print(file_name)
  #Analysis_of_results(df)
  degradable_percent(df,file_name)

def Analysis_of_results(df):
  print('========================================================================================')
  print(df.head())
  print('----------------------------------------------------------------------------------------')
  print(df.describe())
  print('----------------------------------------------------------------------------------------')
  print(df.info())
  print('----------------------------------------------------------------------------------------')
  print(df.isnull().sum())
  print(df.shape)
  print('========================================================================================')
  degradable_percent(df)



```

```python
def degradable_percent(df,file_name):
  bio_degradable=['plant','animal_fish','animal_starfish','animal_shells',
                  'animal_crab','animal_eel',' animal_etc','trash_fabric',
                  'trash_paper','trash_rubber','trash_wood']
  df['bio_degradable'] = df['label'].apply(lambda x: 1 if x in bio_degradable else 0)
  visualize(df,file_name)
def visualize(df,file_name):
    # Count the occurrences of each type
    counts = df['bio_degradable'].value_counts()
    labels=['Non-BioDegradable' if i==0 else 'BioDegradable' for i in counts.index]

    plt.figure(figsize=(20, 10))
    plt.title(f'{file_name}',size=20)
    plt.pie(counts, labels=labels, autopct='%1.1f%%')
    plt.suptitle('Biodegradable vs Non-Biodegradable Waste')
    plt.legend()
    plt.savefig(f'{file_name}.png')
    plt.show()
```

```python
for i in os.listdir():
  if i.endswith('.csv'):
    read_csv(i)
```
